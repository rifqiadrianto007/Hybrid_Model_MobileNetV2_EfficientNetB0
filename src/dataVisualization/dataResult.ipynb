## 1. Radar Chart Tiap Model

```python
import numpy as np
import matplotlib.pyplot as plt

models = ['MobileNetV2', 'EfficientNetB0', 'Fusion Model']

accuracy = [0.8792, 0.9250, 0.9208]
precision = [0.8789, 0.9264, 0.9213]
recall = [0.8792, 0.9250, 0.9208]
f1_score = [0.8789, 0.9253, 0.9205]
auc = [0.9786, 0.9802, 0.9872]

# Dibalik agar waktu lebih kecil = nilai lebih baik
training_efficiency = [1-(190.3508 / 220),
                       1-(216.9797 / 220),
                       1-(174.4734 / 220)]

categories = [
    'Accuracy',
    'Precision',
    'Recall',
    'F1-Score',
    'AUC',
    'Training Efficiency'
]

data = [
    [accuracy[0], precision[0], recall[0], f1_score[0], auc[0], training_efficiency[0]],
    [accuracy[1], precision[1], recall[1], f1_score[1], auc[1], training_efficiency[1]],
    [accuracy[2], precision[2], recall[2], f1_score[2], auc[2], training_efficiency[2]]
]

N = len(categories)

angles = np.linspace(0, 2 * np.pi, N, endpoint = False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize = (8,8), subplot_kw = dict(polar = True))

for i in range(len(models)) :
    values = data[i]
    values += values[:1]
    ax.plot(angles, values, linewidth = 2, label = models[i])
    ax.fill(angles, values, alpha = 0.1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)

plt.title('Radar Chart Perbandingan Performa Model')
plt.legend(loc = 'upper right')
plt.savefig('radar_chart_model.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```

## 2. Scatter Plot Accuracy vs Training Time

```python
import matplotlib.pyplot as plt

models = ['MobileNetV2', 'EfficientNetB0', 'Fusion Model']

training_time = [190.3508, 216.9797, 174.4734]
accuracy = [0.8792, 0.9250, 0.9208]

plt.figure(figsize = (8,6))

for i in range(len(models)) :
    plt.scatter(training_time[i], accuracy[i], s = 250)
    plt.text(training_time[i]+1,
             accuracy[i],
             models[i],
             fontsize = 10)

plt.xlabel('Training Time (Seconds)')
plt.ylabel('Accuracy')
plt.title('Accuracy vs Training Time')
plt.grid(True)
plt.savefig('accuracy_vs_training_time.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```

## 3. Epoch Konvergensi

```python
import matplotlib.pyplot as plt

models = ['MobileNetV2', 'EfficientNetB0', 'Fusion Model']
epochs = [33, 32, 18]

plt.figure(figsize = (8,6))

bars = plt.bar(models, epochs)

for bar in bars :
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.5,
        f'{height}',
        ha = 'center'
    )

plt.ylabel('Total Epoch')
plt.title('Perbandingan Epoch Konvergensi')
plt.savefig('epoch_convergence.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```

## 4. Heatmap Gap Train vs Val Accuracy

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    'Gap Accuracy': [0.080, 0.075, 0.057]
}

index = ['MobileNetV2', 'EfficientNetB0', 'Fusion Model']

df = pd.DataFrame(data, index = index)

plt.figure(figsize = (6,4))

sns.heatmap(
    df,
    annot = True,
    cmap = 'Reds',
    fmt = '.3f'
)

plt.title('Gap Train vs Validation Accuracy')
plt.savefig('gap_accuracy_heatmap.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```
